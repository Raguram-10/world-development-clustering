```python
def outlier_capping(df,column):
    Q1= df[column].quantile(0.25)
    Q3= df[column].quantile(0.75)
    IQR= Q3-Q1
    lower_extreme= Q1-1.5*IQR
    upper_extreme= Q3+1.5*IQR
    df[column]= df[column].apply(lambda x: lower_extreme if x<lower_extreme else upper_extreme if x>upper_extreme else x)
for col in df.select_dtypes(include=['int','float']).columns:
    outlier_capping(df,col)
```

#KMEANS CLUSTERING

```python
# Convert numeric-looking columns into numerical values

for col in df.columns:
    if col != 'Country':
        df[col] = (
            df[col]
            .astype(str)
            .str.replace('$', '', regex=False)
            .str.replace(',', '', regex=False)
            .str.replace('%', '', regex=False)
            .str.strip()
        )
        df[col] = pd.to_numeric(df[col], errors='coerce')

df.dtypes
```

```python
df.dtypes
```

```python
# Select numerical features for clustering
X = df.select_dtypes(include='number').copy()

X.shape
```

```python
# Standardize the features
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

X_scaled.shape
```

```python
from sklearn.cluster import KMeans

inertia = []

for k in range(2, 11):
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X_scaled)
    inertia.append(kmeans.inertia_)

inertia
```

```python
import matplotlib.pyplot as plt

k_values = range(2, 11)

plt.figure(figsize=(8, 5))
plt.plot(k_values, inertia, marker='o')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Inertia')
plt.title('Elbow Method for Optimal K')
plt.xticks(k_values)
plt.grid(True)
plt.show()
```

```python
from sklearn.metrics import silhouette_score

silhouette_scores = []

for k in range(2, 11):
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = kmeans.fit_predict(X_scaled)
    score = silhouette_score(X_scaled, labels)
    silhouette_scores.append(score)

silhouette_scores
```

```python
# Final K-Means model with 4 clusters

kmeans_final = KMeans(
    n_clusters=4,
    random_state=42,
    n_init=10
)

kmeans_labels = kmeans_final.fit_predict(X_scaled)

kmeans_labels
```

```python
df['KMeans_Cluster'] = kmeans_labels

df[['Country', 'KMeans_Cluster']].head(10)
```

```python
df['KMeans_Cluster'].value_counts().sort_index()
```

```python
cluster_summary = df.groupby('KMeans_Cluster')[X.columns].mean()

cluster_summary
```

```python
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt

pca = PCA(n_components=2)

X_pca = pca.fit_transform(X_scaled)

plt.figure(figsize=(10, 6))

plt.scatter(
    X_pca[:, 0],
    X_pca[:, 1],
    c=df['KMeans_Cluster'],
    cmap='viridis',
    s=30
)

plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title('K-Means Clustering - PCA Visualization')
plt.colorbar(label='Cluster')
plt.show()
```

```python
# Final K-Means model with optimal K = 4

kmeans_final = KMeans(
    n_clusters=4,
    random_state=42,
    n_init=10
)

kmeans_labels = kmeans_final.fit_predict(X_scaled)

df['KMeans_Cluster'] = kmeans_labels

df[['Country', 'KMeans_Cluster']].head(10)
```

```python
df['KMeans_Cluster'] = kmeans_labels
```

```python
df['KMeans_Cluster'].value_counts().sort_index()
```

```python
cluster_profile = df.groupby('KMeans_Cluster')[X.columns].mean()

cluster_profile
```

```python
import matplotlib.pyplot as plt

cluster_profile.T.plot(
    kind='bar',
    figsize=(16, 7)
)

plt.title('K-Means Cluster Profiles')
plt.xlabel('Development Indicators')
plt.ylabel('Average Value')
plt.xticks(rotation=90)
plt.legend(title='Cluster')
plt.tight_layout()
plt.show()
```

```python
from sklearn.metrics import silhouette_score

kmeans_silhouette = silhouette_score(
    X_scaled,
    kmeans_labels
)

print("K-Means Silhouette Score:", kmeans_silhouette)
```

HIERARCHICAL CLUSTERING

```python
from sklearn.cluster import AgglomerativeClustering

# Hierarchical Clustering with 4 clusters
hierarchical = AgglomerativeClustering(
    n_clusters=4,
    linkage='ward'
)

hierarchical_labels = hierarchical.fit_predict(X_scaled)

df['Hierarchical_Cluster'] = hierarchical_labels

df[['Country', 'Hierarchical_Cluster']].head(10)
```

```python
from sklearn.metrics import silhouette_score

hierarchical_silhouette = silhouette_score(
    X_scaled,
    hierarchical_labels
)

print("Hierarchical Clustering Silhouette Score:",
      hierarchical_silhouette)
```

```python
# Store Hierarchical Clustering results
df['Hierarchical_Cluster'] = hierarchical_labels

# Show cluster counts
df['Hierarchical_Cluster'].value_counts().sort_index()
```

```python
# Hierarchical cluster profile
hierarchical_profile = df.groupby('Hierarchical_Cluster')[X.columns].mean()

hierarchical_profile
```

```python
from scipy.cluster.hierarchy import dendrogram, linkage
import matplotlib.pyplot as plt

# Create linkage matrix
Z = linkage(X_scaled, method='ward')

# Plot dendrogram
plt.figure(figsize=(12, 6))

dendrogram(
    Z,
    truncate_mode='lastp',
    p=30
)

plt.title('Hierarchical Clustering Dendrogram')
plt.xlabel('Cluster Groups')
plt.ylabel('Distance')
plt.show()
```

DBSCAN

```python
from sklearn.cluster import DBSCAN

# DBSCAN clustering
dbscan = DBSCAN(eps=2, min_samples=10)

dbscan_labels = dbscan.fit_predict(X_scaled)

df['DBSCAN_Cluster'] = dbscan_labels

df[['Country', 'DBSCAN_Cluster']].head(10)
```

```python
df['DBSCAN_Cluster'].value_counts().sort_index()
```

```python
from sklearn.metrics import silhouette_score

# Exclude noise points (-1)
mask = dbscan_labels != -1

if len(set(dbscan_labels[mask])) > 1:
    dbscan_silhouette = silhouette_score(
        X_scaled[mask],
        dbscan_labels[mask]
    )
    print("DBSCAN Silhouette Score:", dbscan_silhouette)
else:
    print("Silhouette score cannot be calculated because DBSCAN produced fewer than 2 clusters.")
```

```python
# PCA visualization for DBSCAN, GMM and Mean Shift

from sklearn.decomposition import PCA
import matplotlib.pyplot as plt

# Reduce the scaled data to 2 dimensions
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

# Create one graph for DBSCAN
plt.figure(figsize=(8, 6))
plt.scatter(X_pca[:, 0], X_pca[:, 1], c=dbscan_labels, cmap='viridis')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title('DBSCAN Clustering')
plt.colorbar(label='Cluster')
plt.show()
```

GMM

```python
# Gaussian Mixture Model

from sklearn.mixture import GaussianMixture

gmm = GaussianMixture(
    n_components=4,
    random_state=42
)

gmm_labels = gmm.fit_predict(X_scaled)

df['GMM_Cluster'] = gmm_labels

df[['Country', 'GMM_Cluster']].head(10)
```

```python
df['GMM_Cluster'].value_counts().sort_index()
```

```python
from sklearn.metrics import silhouette_score

gmm_silhouette = silhouette_score(
    X_scaled,
    gmm_labels
)

print("GMM Silhouette Score:", gmm_silhouette)
```

```python
# GMM Clustering Visualization

plt.figure(figsize=(8, 6))
plt.scatter(X_pca[:, 0], X_pca[:, 1], c=gmm_labels, cmap='viridis')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title('Gaussian Mixture Model Clustering')
plt.colorbar(label='Cluster')
plt.show()
```

MEAN SHIFT

```python
from sklearn.cluster import MeanShift, estimate_bandwidth

# Estimate a suitable bandwidth
bandwidth = estimate_bandwidth(
    X_scaled,
    quantile=0.2,
    n_samples=500
)

print("Estimated Bandwidth:", bandwidth)

# Mean Shift clustering
mean_shift = MeanShift(bandwidth=bandwidth)

mean_shift_labels = mean_shift.fit_predict(X_scaled)

# Add cluster labels
df['MeanShift_Cluster'] = mean_shift_labels

# Check clusters
df['MeanShift_Cluster'].value_counts().sort_index()
```

```python
df['MeanShift_Cluster'].nunique()
```

```python
# Mean Shift Clustering Visualization

plt.figure(figsize=(8, 6))
plt.scatter(X_pca[:, 0], X_pca[:, 1], c=mean_shift_labels, cmap='viridis')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title('Mean Shift Clustering')
plt.colorbar(label='Cluster')
plt.show()
```

```python
model_comparison = pd.DataFrame({
    'Model': [
        'K-Means',
        'Hierarchical',
        'DBSCAN',
        'GMM',
        'Mean Shift'
    ],
    'Silhouette Score': [
        kmeans_silhouette,
        hierarchical_silhouette,
        dbscan_silhouette,
        gmm_silhouette,
        None
    ]
})

model_comparison
```

```python
# Final Comparison of Clustering Models

models = [
    'K-Means',
    'Hierarchical',
    'DBSCAN',
    'GMM'
]

silhouette_scores = [
    kmeans_silhouette,
    hierarchical_silhouette,
    dbscan_silhouette,
    gmm_silhouette
]

plt.figure(figsize=(10, 6))

plt.bar(models, silhouette_scores)

plt.xlabel('Clustering Models')
plt.ylabel('Silhouette Score')
plt.title('Comparison of Clustering Models')

plt.axhline(0, linewidth=0.8)

plt.show()
```

Among the five clustering techniques, K-Means performed best with a Silhouette Score of 0.2666, indicating comparatively better-defined and separated clusters. Hierarchical Clustering and GMM produced lower positive scores, while DBSCAN produced a negative score. Mean Shift assigned all observations to a single cluster, making silhouette-based evaluation not applicable. Therefore, K-Means was selected as the best-performing clustering model for this dataset.”
